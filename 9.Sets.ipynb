SETS

Sets are a built-in data type in Python used to store collection of unique items.They are unordered,meaning that the elements do not follow a specific order,and they do not allow duplicate elements. Sets are useful for membership tests,eliminating duplicate entries, and performing mathematical set operations like union, intersection,difference,and symmetric difference.

Union in Sets
Union combines elements of two sets and removes duplicate values.
Intersection in Sets
Intersection returns elements that are common to both sets.
Difference in Sets
Difference returns elements present in one set but not in another set.

```python
### Create a set
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set=set()
print(type(my_empty_set))
```

```python
my_empty_set=set([1,2,3,4,5])
print(my_empty_set)
```

```python
my_empty_set=set([1,2,3,4,4,5,1])
print(my_empty_set)
```

```python
##Basic set operation
##Adding and removing elements
my_set.add(10)
print(my_set)
```

```python
##Removing a set
my_set.remove(5)
print(my_set)
```

```python
#POP method
removed_element=my_set.pop
print(removed_element)
print(my_set)

```

```python
## MATHEMATICAL OPERATIONS

set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

## UNION
union_set=set1.union(set2)
print(union_set)

##INTERSECTION 
intersection_set=set1.intersection(set2)
print(intersection_set)
```

```python
#Counting unique words in text

text="In this tutorial we are discussig about sets"
words=text.split()

#Convert list of words

unique_words=set(words)
print(unique_words)
print(len(unique_words))
```
